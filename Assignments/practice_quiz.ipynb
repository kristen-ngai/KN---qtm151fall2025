```python
import pandas as pd

# Read CSV file
df = pd.read_csv("data.csv")  # Replace with your file name

# Filter rows where 'age' column is greater than 30
filtered_df = df[df["age"] > 30]

# Print number of rows satisfying the condition
print(len(filtered_df))
```

```python
# Define the function
def categorize_income(income):
    if income >= 50000:
        return "High"
    else:
        return "Low"

# Assume df is a DataFrame with an 'income' column
df["income_category"] = df["income"].apply(categorize_income)

# Optional: View the result
print(df.head())
```

```python
import numpy as np

# Add a new column with random numbers from uniform distribution
df["random_var"] = np.random.rand(len(df))

# Create two groups based on condition (e.g., < 0.5 and ≥ 0.5)
group1 = df[df["random_var"] < 0.5]
group2 = df[df["random_var"] >= 0.5]

print(f"Group 1 size: {len(group1)}")
print(f"Group 2 size: {len(group2)}")
```

```python
import matplotlib.pyplot as plt

# Example scatter plot: plot 'age' vs 'income' for each group
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Group 1
axes[0].scatter(group1["age"], group1["income"], color='blue')
axes[0].set_title("Group 1: Age vs Income")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Income")

# Group 2
axes[1].scatter(group2["age"], group2["income"], color='green')
axes[1].set_title("Group 2: Age vs Income")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Income")

plt.tight_layout()
plt.show()
```
