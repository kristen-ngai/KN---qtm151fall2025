```python
#6. Numerical derivative of sine and comparison with cosine
# Create DataFrame
time = np.linspace(0, 2*np.pi, 100)
df = pd.DataFrame({
    'time': time,
    'sin': np.sin(time),
    'cos': np.cos(time)
})

# Compute numerical derivative of sin
df['d_sin_dt'] = df['sin'].diff() / df['time'].diff()

# Plot cos and numerical derivative
plt.figure(figsize=(10,6))
plt.plot(df['time'], df['cos'], label='cos(t)')
plt.plot(df['time'], df['d_sin_dt'], '--', label='d(sin)/dt')
plt.xlabel('Time')
plt.ylabel('Value')
plt.title('cos(t) vs Numerical Derivative of sin(t)')
plt.legend()
plt.show()
```
